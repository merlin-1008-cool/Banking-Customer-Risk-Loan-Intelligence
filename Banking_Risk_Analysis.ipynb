```python
# Multiple Loans per Customer Analysis

customer_loan_exposure = loans.groupby('customer_id').agg(
    total_loans=('loan_id', 'count'),
    total_loan_amount=('loan_amount', 'sum'),
    total_outstanding=('outstanding_amount', 'sum')
).reset_index()

multiple_loan_customers = customer_loan_exposure[
    customer_loan_exposure['total_loans'] > 1
].copy()

multiple_loan_summary = pd.DataFrame({
    'Metric': [
        'Total Customers with Multiple Loans',
        'Total Loans Held by These Customers',
        'Total Loan Amount',
        'Total Outstanding Amount'
    ],
    'Value': [
        multiple_loan_customers['customer_id'].nunique(),
        multiple_loan_customers['total_loans'].sum(),
        multiple_loan_customers['total_loan_amount'].sum(),
        multiple_loan_customers['total_outstanding'].sum()
    ]
})

# Format output to display full numeric values
multiple_loan_summary['Value'] = multiple_loan_summary.apply(
    lambda row: f"{int(row['Value']):,}"
    if row['Metric'] in [
        'Total Customers with Multiple Loans',
        'Total Loans Held by These Customers'
    ]
    else f"{row['Value']:,.2f}",
    axis=1
)

display(multiple_loan_summary)
```

```python
# Customer Default Risk by Number of Loans

# Count loans for each customer
customer_loan_count = loans.groupby('customer_id').agg(
    total_loans=('loan_id', 'count')
).reset_index()

# Identify customers who have at least one defaulted loan
customer_default = loans.groupby('customer_id').agg(
    has_default=('loan_status', lambda x: (x == 'Default').any())
).reset_index()

# Combine loan count and default information
customer_risk = customer_loan_count.merge(
    customer_default,
    on='customer_id'
)

# Group customers by number of loans
customer_risk['loan_category'] = customer_risk['total_loans'].apply(
    lambda x: 'Single Loan' if x == 1 else 'Multiple Loans'
)

# Calculate default risk for each category
default_risk_by_loan_count = customer_risk.groupby('loan_category').agg(
    total_customers=('customer_id', 'count'),
    customers_with_default=('has_default', 'sum')
).reset_index()

# Calculate default rate
default_risk_by_loan_count['default_rate'] = (
    default_risk_by_loan_count['customers_with_default']
    / default_risk_by_loan_count['total_customers']
    * 100
)

# Display results
default_risk_by_loan_count
```

```python
# Customer Default Risk by Outstanding Amount

# Step 1: Calculate total outstanding amount for each customer
customer_outstanding = loans.groupby('customer_id').agg(
    total_outstanding=('outstanding_amount', 'sum')
).reset_index()

# Step 2: Identify customers with at least one defaulted loan
customer_defaults = loans.groupby('customer_id').agg(
    has_default=('loan_status', lambda x: (x == 'Default').any())
).reset_index()

# Step 3: Combine outstanding amount and default information
customer_risk_outstanding = customer_outstanding.merge(
    customer_defaults,
    on='customer_id'
)

# Step 4: Divide customers into five groups based on outstanding amount
customer_risk_outstanding['outstanding_group'] = pd.qcut(
    customer_risk_outstanding['total_outstanding'],
    q=5,
    labels=['Very Low', 'Low', 'Medium', 'High', 'Very High'],
    duplicates='drop'
)

# Step 5: Calculate default rate for each group
default_by_outstanding = customer_risk_outstanding.groupby(
    'outstanding_group',
    observed=False
).agg(
    total_customers=('customer_id', 'count'),
    customers_with_default=('has_default', 'sum')
).reset_index()

# Step 6: Calculate default rate percentage
default_by_outstanding['default_rate'] = (
    default_by_outstanding['customers_with_default']
    / default_by_outstanding['total_customers']
    * 100
).round(2)

# Step 7: Display results
default_by_outstanding
```

```python
# ============================================================
# BANKING CUSTOMER RISK & LOAN INTELLIGENCE
# FINAL LOAN DEFAULT PREDICTION MODEL
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ------------------------------------------------------------
# 1. PREPARE DATA
# ------------------------------------------------------------

data = ml_data.copy()

# Keep only legitimate target values
data = data[data["default_flag"].isin([0, 1])].copy()
data["default_flag"] = data["default_flag"].astype(int)

# Use only features available for risk assessment.
# Exclude loan_status, IDs, and dates to prevent leakage.
features = [
    "loan_type",
    "loan_amount",
    "interest_rate",
    "age",
    "gender",
    "occupation",
    "annual_income"
]

X = data[features].copy()
y = data["default_flag"]

numeric_features = [
    "loan_amount",
    "interest_rate",
    "age",
    "annual_income"
]

categorical_features = [
    "loan_type",
    "gender",
    "occupation"
]

print("Dataset shape:", data.shape)
print("Total loans:", len(data))
print("Defaulted loans:", int(y.sum()))
print("Default rate:", round(y.mean() * 100, 2), "%")

# ------------------------------------------------------------
# 2. PREPROCESSING
# ------------------------------------------------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# ------------------------------------------------------------
# 3. SPLIT DATA
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# ------------------------------------------------------------
# 4. TRAIN RANDOM FOREST
# ------------------------------------------------------------

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=500,
        max_depth=5,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

model.fit(X_train, y_train)

# ------------------------------------------------------------
# 5. EVALUATE ON UNSEEN TEST DATA
# ------------------------------------------------------------

probabilities = model.predict_proba(X_test)[:, 1]

# Default threshold: 0.50
predictions = (probabilities >= 0.50).astype(int)

print("\n" + "=" * 60)
print("FINAL MODEL EVALUATION")
print("=" * 60)

print("ROC-AUC:",
      round(roc_auc_score(y_test, probabilities), 4))

print("Average Precision:",
      round(average_precision_score(y_test, probabilities), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    predictions,
    labels=[0, 1],
    target_names=["Non-Default", "Default"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, predictions))

# ------------------------------------------------------------
# 6. CROSS-VALIDATION ON TRAINING DATA
# ------------------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    model,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "roc_auc": "roc_auc",
        "average_precision": "average_precision",
        "recall": "recall"
    },
    n_jobs=1
)

print("\n" + "=" * 60)
print("5-FOLD CROSS-VALIDATION")
print("=" * 60)

print(
    "Mean ROC-AUC:",
    round(cv_results["test_roc_auc"].mean(), 4)
)

print(
    "Mean Average Precision:",
    round(cv_results["test_average_precision"].mean(), 4)
)

print(
    "Mean Recall:",
    round(cv_results["test_recall"].mean(), 4)
)

# ------------------------------------------------------------
# 7. CONFUSION MATRIX VISUALIZATION
# ------------------------------------------------------------

ConfusionMatrixDisplay.from_predictions(
    y_test,
    predictions,
    labels=[0, 1],
    display_labels=["Non-Default", "Default"]
)

plt.title("Loan Default Prediction - Random Forest")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. SAVE THE TRAINED MODEL
# ------------------------------------------------------------

joblib.dump(
    model,
    "banking_loan_default_model.pkl"
)

print("\nModel saved as banking_loan_default_model.pkl")

print("\nImportant:")
print("This model uses the existing synthetic default labels.")
print("Its predictive performance must be judged from the")
print("test results and cross-validation above.")
```

```python
import pandas as pd
import os

# Locate existing project data
project_folder = os.path.dirname(os.getcwd())
data_folder = os.path.join(project_folder, "data")

# Load existing datasets
loans = pd.read_csv(os.path.join(data_folder, "Loans.csv"))
customers = pd.read_csv(os.path.join(data_folder, "customers.csv"))

# Create the target variable
loans["default_flag"] = (loans["loan_status"] == "Default").astype(int)

# Combine loan and customer information
ml_data = loans.merge(
    customers,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

# Display results
print("ML dataset shape:", ml_data.shape)
print("\nDefault flag distribution:")
print(ml_data["default_flag"].value_counts())

print("\nDefault rate:", round(ml_data["default_flag"].mean() * 100, 2), "%")
print("\nAvailable features:")
print(ml_data.columns.tolist())

display(ml_data.head())
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    average_precision_score
)

# 1. Select predictors and target
target = "default_flag"

features = [
    "loan_type",
    "loan_amount",
    "interest_rate",
    "outstanding_amount",
    "age",
    "gender",
    "city",
    "occupation",
    "annual_income"
]

X = ml_data[features].copy()
y = ml_data[target].copy()

# 2. Identify numeric and categorical columns
numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X.select_dtypes(exclude=["number"]).columns.tolist()

# 3. Prepare preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# 4. Split data while preserving the default proportion
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 5. Build the model
model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        max_depth=6,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# 6. Train once
model.fit(X_train, y_train)

# 7. Evaluate on the untouched test set
predictions = model.predict(X_test)
probabilities = model.predict_proba(X_test)[:, 1]

print("=" * 55)
print("BANKING LOAN DEFAULT PREDICTION MODEL")
print("=" * 55)

print(f"Training records: {len(X_train)}")
print(f"Testing records:  {len(X_test)}")
print(f"Test defaults:    {int(y_test.sum())}")
print(f"Test non-defaults: {int((y_test == 0).sum())}")

print(f"\nAccuracy: {accuracy_score(y_test, predictions):.4f}")
print(f"ROC-AUC: {roc_auc_score(y_test, probabilities):.4f}")
print(f"Average precision: {average_precision_score(y_test, probabilities):.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    predictions,
    target_names=["Non-default", "Default"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, predictions))
```

```python
# ================================================================
# BANKING CUSTOMER RISK & LOAN INTELLIGENCE
# COMPLETE MACHINE LEARNING PIPELINE
# ================================================================

import os
import json
import warnings
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# ------------------------------------------------
# 1. FIND PROJECT AND LOAD EXISTING DATA
# ------------------------------------------------

possible_roots = [
    os.path.dirname(os.getcwd()),
    os.getcwd(),
    os.path.expanduser(
        "~/Desktop/AI/Banking-Customer-Risk-Loan-Intelligence"
    )
]

PROJECT = None

for root in possible_roots:
    if os.path.exists(os.path.join(root, "data", "Loans.csv")):
        PROJECT = root
        break

if PROJECT is None:
    raise FileNotFoundError(
        "Could not locate data/Loans.csv. "
        "Please run this cell from your existing project notebook."
    )

DATA = os.path.join(PROJECT, "data")
OUTPUT = os.path.join(PROJECT, "ml_results")
os.makedirs(OUTPUT, exist_ok=True)

loans = pd.read_csv(os.path.join(DATA, "Loans.csv"))
customers = pd.read_csv(os.path.join(DATA, "customers.csv"))

print("=" * 65)
print("BANKING CUSTOMER RISK & LOAN INTELLIGENCE")
print("=" * 65)
print("Project:", PROJECT)
print("Loans:", loans.shape)
print("Customers:", customers.shape)

# ------------------------------------------------
# 2. VALIDATE AND PREPARE TARGET
# ------------------------------------------------

required_loan_columns = [
    "loan_id", "customer_id", "loan_type", "loan_amount",
    "interest_rate", "loan_date", "loan_status",
    "outstanding_amount"
]

missing = set(required_loan_columns) - set(loans.columns)

if missing:
    raise ValueError(f"Missing required loan columns: {sorted(missing)}")

if loans["loan_id"].duplicated().any():
    raise ValueError("Duplicate loan IDs detected. Investigate before training.")

loans["loan_status"] = loans["loan_status"].astype(str).str.strip()

# Closed and Active loans are treated as non-defaults.
# Default loans are the positive class.
loans = loans[loans["loan_status"].isin(
    ["Active", "Closed", "Default"]
)].copy()

loans["default_flag"] = (
    loans["loan_status"] == "Default"
).astype(int)

if loans["default_flag"].nunique() != 2:
    raise ValueError("The target must contain both default classes.")

customers = customers.drop_duplicates("customer_id")

data = loans.merge(
    customers,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

print("\nTarget distribution:")
print(data["loan_status"].value_counts())
print(f"Overall default rate: {data['default_flag'].mean():.2%}")

# ------------------------------------------------
# 3. FEATURE ENGINEERING
# ------------------------------------------------

# Parse dates and calculate elapsed time.
data["loan_date"] = pd.to_datetime(data["loan_date"], errors="coerce")
data["customer_since"] = pd.to_datetime(
    data["customer_since"], errors="coerce"
)

data["loan_year"] = data["loan_date"].dt.year
data["loan_month"] = data["loan_date"].dt.month

data["customer_tenure_years"] = (
    (data["loan_date"] - data["customer_since"]).dt.days / 365.25
).clip(lower=0)

# Financial ratios.
# These are candidate predictors, not guaranteed signals of default.
data["loan_to_income"] = (
    data["loan_amount"] /
    data["annual_income"].where(data["annual_income"] > 0)
)

data["outstanding_to_income"] = (
    data["outstanding_amount"] /
    data["annual_income"].where(data["annual_income"] > 0)
)

data["outstanding_to_loan"] = (
    data["outstanding_amount"] /
    data["loan_amount"].where(data["loan_amount"] > 0)
)

# Remove invalid numeric values.
data.replace([np.inf, -np.inf], np.nan, inplace=True)

# ------------------------------------------------
# 4. DEFINE PREDICTORS
# ------------------------------------------------

# Exclude identifiers, the target, and loan_status to prevent leakage.
# Note: outstanding_amount may reflect loan repayment progress.
# Its suitability depends on when the prediction is intended to occur.
features = [
    "loan_type",
    "loan_amount",
    "interest_rate",
    "age",
    "gender",
    "city",
    "occupation",
    "annual_income",
    "customer_tenure_years",
    "loan_to_income",
    "outstanding_to_income",
    "outstanding_to_loan",
    "loan_year",
    "loan_month"
]

X = data[features].copy()
y = data["default_flag"].copy()

numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = [
    col for col in X.columns if col not in numeric_features
]

# ------------------------------------------------
# 5. STRATIFIED TRAIN/TEST SPLIT
# ------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Defaults in test set:", int(y_test.sum()))

# ------------------------------------------------
# 6. PREPROCESSING
# ------------------------------------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# ------------------------------------------------
# 7. COMPARE MODELS USING TRAINING DATA ONLY
# ------------------------------------------------

candidates = {
    "Logistic Regression": LogisticRegression(
        class_weight="balanced",
        max_iter=2000,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=6,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        max_depth=6,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []
pipelines = {}

print("\nComparing models with 5-fold cross-validation...")

for name, estimator in candidates.items():
    pipe = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", estimator)
    ])

    scores = cross_val_score(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring="roc_auc",
        n_jobs=1
    )

    pipelines[name] = pipe

    results.append({
        "Model": name,
        "Mean_CV_ROC_AUC": scores.mean(),
        "CV_Std": scores.std()
    })

comparison = pd.DataFrame(results).sort_values(
    "Mean_CV_ROC_AUC",
    ascending=False
).reset_index(drop=True)

print("\nCROSS-VALIDATION RESULTS")
print(comparison.to_string(index=False))

# ------------------------------------------------
# 8. TRAIN SELECTED MODEL
# ------------------------------------------------

best_name = comparison.loc[0, "Model"]
best_model = pipelines[best_name]

best_model.fit(X_train, y_train)

probabilities = best_model.predict_proba(X_test)[:, 1]

# Standard threshold: 0.50.
# Do not tune using the test set.
predictions = (probabilities >= 0.50).astype(int)

# ------------------------------------------------
# 9. FINAL EVALUATION
# ------------------------------------------------

metrics = {
    "selected_model": best_name,
    "accuracy": float(accuracy_score(y_test, predictions)),
    "roc_auc": float(roc_auc_score(y_test, probabilities)),
    "average_precision": float(
        average_precision_score(y_test, probabilities)
    ),
    "default_precision": float(
        precision_score(y_test, predictions, zero_division=0)
    ),
    "default_recall": float(
        recall_score(y_test, predictions, zero_division=0)
    ),
    "default_f1": float(
        f1_score(y_test, predictions, zero_division=0)
    ),
    "test_records": int(len(y_test)),
    "test_defaults": int(y_test.sum()),
    "threshold": 0.50
}

print("\n" + "=" * 65)
print("FINAL TEST RESULTS")
print("=" * 65)

for key, value in metrics.items():
    if isinstance(value, float):
        print(f"{key}: {value:.4f}")
    else:
        print(f"{key}: {value}")

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test,
    predictions,
    labels=[0, 1],
    target_names=["Non-default", "Default"],
    zero_division=0
))

cm = confusion_matrix(y_test, predictions, labels=[0, 1])

print("CONFUSION MATRIX")
print(pd.DataFrame(
    cm,
    index=["Actual Non-default", "Actual Default"],
    columns=["Predicted Non-default", "Predicted Default"]
))

# ------------------------------------------------
# 10. FEATURE IMPORTANCE
# ------------------------------------------------

# Permutation importance evaluates how much test ROC-AUC changes
# when a feature is shuffled. It is diagnostic, not causal evidence.
from sklearn.inspection import permutation_importance

permutation = permutation_importance(
    best_model,
    X_test,
    y_test,
    scoring="roc_auc",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

importance = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance_Mean": permutation.importances_mean,
    "Importance_Std": permutation.importances_std
}).sort_values("Importance_Mean", ascending=False)

print("\nTOP FEATURES BY PERMUTATION IMPORTANCE")
print(importance.head(10).to_string(index=False))

# ------------------------------------------------
# 11. SAVE MODEL AND RESULTS
# ------------------------------------------------

joblib.dump(
    best_model,
    os.path.join(OUTPUT, "banking_default_model.joblib")
)

comparison.to_csv(
    os.path.join(OUTPUT, "model_comparison.csv"),
    index=False
)

importance.to_csv(
    os.path.join(OUTPUT, "feature_importance.csv"),
    index=False
)

with open(
    os.path.join(OUTPUT, "evaluation_metrics.json"),
    "w"
) as f:
    json.dump(metrics, f, indent=4)

test_results = pd.DataFrame({
    "actual_default": y_test.to_numpy(),
    "predicted_default": predictions,
    "predicted_default_probability": probabilities
})

test_results.to_csv(
    os.path.join(OUTPUT, "test_predictions.csv"),
    index=False
)

print("\n" + "=" * 65)
print("PIPELINE COMPLETED")
print("=" * 65)
print("Saved model and reports to:", OUTPUT)
print("1. banking_default_model.joblib")
print("2. model_comparison.csv")
print("3. feature_importance.csv")
print("4. evaluation_metrics.json")
print("5. test_predictions.csv")
```

```python
# ============================================================
# FINAL MODEL EXPERIMENT: BANKING LOAN DEFAULT PREDICTION
# Reuses the existing ml_data dataframe
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
    GradientBoostingClassifier
)
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# 1. Prepare target
data = ml_data.copy()
data = data[data["loan_status"].isin(["Active", "Closed", "Default"])].copy()
data["default_flag"] = (data["loan_status"] == "Default").astype(int)

# 2. Create features available from loan/customer records
data["loan_date"] = pd.to_datetime(data["loan_date"], errors="coerce")
data["customer_since"] = pd.to_datetime(data["customer_since"], errors="coerce")

data["customer_tenure_years"] = (
    (data["loan_date"] - data["customer_since"]).dt.days / 365.25
).clip(lower=0)

data["loan_to_income"] = (
    data["loan_amount"] / data["annual_income"].where(data["annual_income"] > 0)
)

data["interest_to_income"] = (
    data["loan_amount"] * data["interest_rate"] / 100
) / data["annual_income"].where(data["annual_income"] > 0)

data["loan_year"] = data["loan_date"].dt.year
data["loan_month"] = data["loan_date"].dt.month

# Exclude outstanding_amount to avoid potential post-origination leakage.
features = [
    "loan_type",
    "loan_amount",
    "interest_rate",
    "age",
    "gender",
    "city",
    "occupation",
    "annual_income",
    "customer_tenure_years",
    "loan_to_income",
    "interest_to_income",
    "loan_year",
    "loan_month"
]

X = data[features].replace([np.inf, -np.inf], np.nan)
y = data["default_flag"]

numeric = X.select_dtypes(include="number").columns.tolist()
categorical = [c for c in X.columns if c not in numeric]

# 3. One fixed stratified test set
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scale", StandardScaler())
    ]), numeric),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encode", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical)
])

# 4. Compare candidate algorithms using training data only
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=3000, class_weight="balanced", random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=400, max_depth=6, min_samples_leaf=10,
        class_weight="balanced", random_state=42, n_jobs=-1
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=400, max_depth=6, min_samples_leaf=10,
        class_weight="balanced", random_state=42, n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100, learning_rate=0.03,
        max_depth=2, random_state=42
    )
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
pipelines = {}
cv_results = []

for name, estimator in models.items():
    pipe = Pipeline([
        ("prep", preprocessor),
        ("model", estimator)
    ])

    scores = cross_val_score(
        pipe, X_train, y_train,
        scoring="roc_auc", cv=cv, n_jobs=1
    )

    pipelines[name] = pipe
    cv_results.append({
        "Model": name,
        "CV_ROC_AUC_Mean": scores.mean(),
        "CV_ROC_AUC_Std": scores.std()
    })

comparison = pd.DataFrame(cv_results).sort_values(
    "CV_ROC_AUC_Mean", ascending=False
).reset_index(drop=True)

print("\nCROSS-VALIDATED MODEL COMPARISON")
print(comparison.to_string(index=False))

# 5. Fit the best model according to cross-validation
best_name = comparison.iloc[0]["Model"]
best_model = pipelines[best_name]
best_model.fit(X_train, y_train)

prob = best_model.predict_proba(X_test)[:, 1]
pred = (prob >= 0.50).astype(int)

print("\nSELECTED MODEL:", best_name)
print("Test accuracy:", round(accuracy_score(y_test, pred), 4))
print("Balanced accuracy:", round(balanced_accuracy_score(y_test, pred), 4))
print("Test ROC-AUC:", round(roc_auc_score(y_test, prob), 4))
print("Average precision:", round(average_precision_score(y_test, prob), 4))
print("Default precision:", round(precision_score(y_test, pred, zero_division=0), 4))
print("Default recall:", round(recall_score(y_test, pred, zero_division=0), 4))
print("Default F1:", round(f1_score(y_test, pred, zero_division=0), 4))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test, pred,
    labels=[0, 1],
    target_names=["Non-default", "Default"],
    zero_division=0
))

print("CONFUSION MATRIX")
print(confusion_matrix(y_test, pred))

# 6. Save this experiment separately, preserving previous results
project = os.path.dirname(os.getcwd())
output = os.path.join(project, "ml_results")
os.makedirs(output, exist_ok=True)

joblib.dump(best_model, os.path.join(output, "banking_default_model_v2.joblib"))
comparison.to_csv(os.path.join(output, "model_comparison_v2.csv"), index=False)

pd.DataFrame({
    "actual_default": y_test.to_numpy(),
    "predicted_default": pred,
    "default_probability": prob
}).to_csv(os.path.join(output, "test_predictions_v2.csv"), index=False)

print("\nSaved new experiment to:", output)
print("Previous model files remain unchanged.")
```

```python
# ============================================================
# BANKING CUSTOMER RISK & LOAN INTELLIGENCE
# Improved Synthetic Default Prediction Experiment
# ============================================================

from pathlib import Path
import json
import warnings
import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. FIND THE PROJECT AND LOAD THE ORIGINAL DATA
# ------------------------------------------------------------

possible_roots = [
    Path.cwd(),
    Path.cwd().parent,
    Path.home() / "Desktop" / "AI" / "Banking-Customer-Risk-Loan-Intelligence"
]

project_root = None

for root in possible_roots:
    if (
        (root / "data" / "customers.csv").exists()
        and (root / "data" / "Loans.csv").exists()
    ):
        project_root = root
        break

if project_root is None:
    raise FileNotFoundError(
        "Could not locate data/customers.csv and data/Loans.csv. "
        "Open this notebook from your Banking-Customer-Risk-Loan-Intelligence project."
    )

customers = pd.read_csv(project_root / "data" / "customers.csv")
loans = pd.read_csv(project_root / "data" / "Loans.csv")

print("Project folder:", project_root)
print("Customers:", customers.shape)
print("Loans:", loans.shape)

# ------------------------------------------------------------
# 2. JOIN CUSTOMER AND LOAN INFORMATION
# ------------------------------------------------------------

df = loans.merge(
    customers,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

required_columns = [
    "loan_type", "loan_amount", "interest_rate",
    "age", "annual_income"
]

missing_columns = [c for c in required_columns if c not in df.columns]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

# Convert numeric columns safely
for col in ["loan_amount", "interest_rate", "age", "annual_income"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove rows missing essential inputs
df = df.dropna(
    subset=["loan_type", "loan_amount", "interest_rate", "age", "annual_income"]
).copy()

# Avoid division by zero
df["annual_income"] = df["annual_income"].clip(lower=1)

# ------------------------------------------------------------
# 3. ENGINEER RISK FEATURES
# ------------------------------------------------------------

df["loan_to_income"] = (
    df["loan_amount"] / df["annual_income"]
)

df["income_per_month"] = (
    df["annual_income"] / 12
)

# A simple age-risk indicator; it is an assumption, not a fact
df["age_risk"] = (
    (df["age"] < 25) | (df["age"] > 60)
).astype(int)

# Loan-type risk weights are explicit synthetic assumptions
loan_type_risk = {
    "Personal": 0.45,
    "Business": 0.35,
    "Auto": 0.10,
    "Education": 0.00,
    "Home": -0.20
}

df["loan_type_risk"] = (
    df["loan_type"].astype(str).map(loan_type_risk).fillna(0)
)

# ------------------------------------------------------------
# 4. GENERATE A SEPARATE SYNTHETIC DEFAULT TARGET
# ------------------------------------------------------------
# IMPORTANT:
# These are simulated labels generated from explicit assumptions.
# They are NOT the original observed loan_status labels.

# Standardize features for stable risk-score construction
interest_z = (
    (df["interest_rate"] - df["interest_rate"].mean())
    / (df["interest_rate"].std() + 1e-9)
).clip(-3, 3)

loan_ratio_z = (
    (df["loan_to_income"] - df["loan_to_income"].mean())
    / (df["loan_to_income"].std() + 1e-9)
).clip(-3, 3)

income_z = (
    (df["annual_income"] - df["annual_income"].mean())
    / (df["annual_income"].std() + 1e-9)
).clip(-3, 3)

# Higher interest and higher loan-to-income increase simulated risk.
# Higher income slightly reduces risk. Age and loan type also contribute.
risk_score = (
    0.95 * interest_z
    + 1.05 * loan_ratio_z
    - 0.40 * income_z
    + 0.35 * df["age_risk"]
    + 0.75 * df["loan_type_risk"]
)

# Calibrate probabilities to an average default probability of 9.5%
target_default_rate = 0.095

def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -30, 30)))

low, high = -20.0, 20.0

for _ in range(100):
    midpoint = (low + high) / 2
    mean_probability = sigmoid(risk_score + midpoint).mean()

    if mean_probability > target_default_rate:
        high = midpoint
    else:
        low = midpoint

intercept = (low + high) / 2
df["synthetic_default_probability"] = sigmoid(risk_score + intercept)

# Fixed seed makes the experiment reproducible
rng = np.random.default_rng(42)

df["synthetic_default_flag"] = rng.binomial(
    1, df["synthetic_default_probability"]
)

# Save a separate dataset; original files are untouched
output_dir = project_root / "ml_results"
output_dir.mkdir(parents=True, exist_ok=True)

dataset_path = output_dir / "synthetic_default_risk_dataset.csv"
df.to_csv(dataset_path, index=False)

print("\nSynthetic dataset created.")
print("Rows:", len(df))
print("Synthetic defaults:", int(df["synthetic_default_flag"].sum()))
print(
    "Synthetic default rate:",
    f"{df['synthetic_default_flag'].mean():.2%}"
)
print("Saved dataset:", dataset_path)

# ------------------------------------------------------------
# 5. SELECT MODEL INPUT FEATURES
# ------------------------------------------------------------

numeric_features = [
    "loan_amount",
    "interest_rate",
    "age",
    "annual_income",
    "loan_to_income",
    "age_risk"
]

categorical_features = ["loan_type"]

X = df[numeric_features + categorical_features].copy()
y = df["synthetic_default_flag"].copy()

# ------------------------------------------------------------
# 6. TRAIN / TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# ------------------------------------------------------------
# 7. COMPARE THREE MODELS
# ------------------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=400,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": HistGradientBoostingClassifier(
        max_iter=150,
        learning_rate=0.06,
        max_leaf_nodes=15,
        l2_regularization=1.0,
        random_state=42
    )
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []
trained_models = {}

for model_name, estimator in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", estimator)
    ])

    cv_scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring="roc_auc",
        n_jobs=-1
    )

    pipeline.fit(X_train, y_train)

    probabilities = pipeline.predict_proba(X_test)[:, 1]
    predictions = (probabilities >= 0.50).astype(int)

    results.append({
        "Model": model_name,
        "CV_ROC_AUC_Mean": cv_scores.mean(),
        "CV_ROC_AUC_Std": cv_scores.std(),
        "Test_Accuracy": accuracy_score(y_test, predictions),
        "Test_Balanced_Accuracy": balanced_accuracy_score(
            y_test, predictions
        ),
        "Test_ROC_AUC": roc_auc_score(y_test, probabilities),
        "Test_Average_Precision": average_precision_score(
            y_test, probabilities
        ),
        "Default_Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Default_Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "Default_F1": f1_score(
            y_test, predictions, zero_division=0
        )
    })

    trained_models[model_name] = pipeline

results_df = pd.DataFrame(results).sort_values(
    "CV_ROC_AUC_Mean",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 65)
print("MODEL COMPARISON")
print("=" * 65)
print(results_df.round(4).to_string(index=False))

results_df.to_csv(
    output_dir / "synthetic_model_comparison.csv",
    index=False
)

# ------------------------------------------------------------
# 8. SAVE AND EVALUATE THE MODEL SELECTED BY CROSS-VALIDATION
# ------------------------------------------------------------

best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]

best_probabilities = best_model.predict_proba(X_test)[:, 1]
best_predictions = (best_probabilities >= 0.50).astype(int)

joblib.dump(
    best_model,
    output_dir / "synthetic_default_risk_model.joblib"
)

evaluation = {
    "experiment": "Synthetic default-risk prediction",
    "target": "synthetic_default_flag",
    "best_model": best_name,
    "training_rows": int(len(X_train)),
    "test_rows": int(len(X_test)),
    "synthetic_default_rate": float(y.mean()),
    "test_accuracy": float(accuracy_score(y_test, best_predictions)),
    "test_balanced_accuracy": float(
        balanced_accuracy_score(y_test, best_predictions)
    ),
    "test_roc_auc": float(
        roc_auc_score(y_test, best_probabilities)
    ),
    "test_average_precision": float(
        average_precision_score(y_test, best_probabilities)
    ),
    "default_precision": float(
        precision_score(y_test, best_predictions, zero_division=0)
    ),
    "default_recall": float(
        recall_score(y_test, best_predictions, zero_division=0)
    ),
    "default_f1": float(
        f1_score(y_test, best_predictions, zero_division=0)
    ),
    "confusion_matrix": confusion_matrix(
        y_test, best_predictions
    ).tolist(),
    "note": (
        "Target labels were simulated from assumed risk relationships. "
        "Results demonstrate a synthetic ML workflow and do not establish "
        "real-world default prediction performance."
    )
}

with open(output_dir / "synthetic_model_evaluation.json", "w") as f:
    json.dump(evaluation, f, indent=4)

predictions_df = X_test.copy()
predictions_df["actual_synthetic_default"] = y_test.values
predictions_df["predicted_default_probability"] = best_probabilities
predictions_df["predicted_synthetic_default"] = best_predictions

predictions_df.to_csv(
    output_dir / "synthetic_test_predictions.csv",
    index=False
)

print("\n" + "=" * 65)
print("BEST MODEL EVALUATION")
print("=" * 65)
print("Selected model:", best_name)
print(f"Accuracy: {evaluation['test_accuracy']:.4f}")
print(f"Balanced accuracy: {evaluation['test_balanced_accuracy']:.4f}")
print(f"ROC-AUC: {evaluation['test_roc_auc']:.4f}")
print(f"Average precision: {evaluation['test_average_precision']:.4f}")
print(f"Default precision: {evaluation['default_precision']:.4f}")
print(f"Default recall: {evaluation['default_recall']:.4f}")
print(f"Default F1-score: {evaluation['default_f1']:.4f}")

print("\nConfusion matrix:")
print(np.array(evaluation["confusion_matrix"]))

print("\nClassification report:")
print(classification_report(
    y_test,
    best_predictions,
    target_names=["Non-default", "Synthetic default"],
    zero_division=0
))

print("\nSaved all experiment files in:")
print(output_dir)
print("\nOriginal customer and loan CSV files were not modified.")
```
