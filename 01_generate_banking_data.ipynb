```python
import pandas as pd
import numpy as np

# Reproducibility
np.random.seed(42)

# Number of customers
n_customers = 10_000

# Customer IDs
customer_ids = [f"C{10001 + i}" for i in range(n_customers)]

# Indian cities
cities = [
    "Bengaluru",
    "Chennai",
    "Mumbai",
    "Hyderabad",
    "Delhi",
    "Pune",
    "Kolkata",
    "Kochi",
    "Coimbatore",
    "Ahmedabad"
]

# Occupations
occupations = [
    "IT Professional",
    "Business Owner",
    "Government Employee",
    "Private Employee",
    "Teacher",
    "Healthcare Professional",
    "Self Employed",
    "Engineer",
    "Consultant",
    "Other"
]

# Generate customer data
customers = pd.DataFrame({
    "customer_id": customer_ids,
    "age": np.random.randint(21, 66, n_customers),
    "gender": np.random.choice(
        ["Male", "Female"],
        size=n_customers,
        p=[0.55, 0.45]
    ),
    "city": np.random.choice(
        cities,
        size=n_customers
    ),
    "occupation": np.random.choice(
        occupations,
        size=n_customers
    ),
    "annual_income": np.random.randint(
        250000,
        2500001,
        n_customers
    ),
    "customer_since": pd.to_datetime(
        np.random.choice(
            pd.date_range("2010-01-01", "2025-12-31"),
            size=n_customers
        )
    )
})

# Display first 5 customers
customers.head()
```

```python
print("Shape:", customers.shape)
print("Columns:", customers.columns.tolist())
print("Missing values:")
print(customers.isnull().sum())
```

```python
customers.to_csv("../data/customers.csv", index=False)
print("customers.csv created successfully!")
```

```python
# Generate account data

np.random.seed(42)

# Number of accounts
n_accounts = 13_000

# Account IDs
account_ids = [f"A{10001 + i}" for i in range(n_accounts)]

# Randomly assign customers
account_customer_ids = np.random.choice(
    customers["customer_id"],
    size=n_accounts
)

# Account types
account_types = np.random.choice(
    ["Savings", "Current"],
    size=n_accounts,
    p=[0.75, 0.25]
)

# Account balances
account_balances = np.round(
    np.random.lognormal(
        mean=11,
        sigma=1.0,
        size=n_accounts
    ),
    2
)

# Account opening dates
account_opening_dates = pd.to_datetime(
    np.random.choice(
        pd.date_range("2012-01-01", "2025-12-31"),
        size=n_accounts
    )
)

# Account status
account_status = np.random.choice(
    ["Active", "Inactive"],
    size=n_accounts,
    p=[0.90, 0.10]
)

accounts = pd.DataFrame({
    "account_id": account_ids,
    "customer_id": account_customer_ids,
    "account_type": account_types,
    "balance": account_balances,
    "account_opening_date": account_opening_dates,
    "status": account_status
})

accounts.head()
```

```python
print("Shape:", accounts.shape)
print("\nColumns:", accounts.columns.tolist())
print("\nMissing values:")
print(accounts.isnull().sum())
```

```python
accounts.to_csv("../data/accounts.csv", index=False)

print("accounts.csv created successfully!")
```

```python
import pandas as pd
import random
from datetime import datetime, timedelta

random.seed(42)

# Load existing accounts
accounts = pd.read_csv("../Data/Accounts.csv")

transactions = []

transaction_types = ["Credit", "Debit", "Transfer"]

for i in range(1, 5001):

    account_id = random.choice(accounts["account_id"].tolist())

    transaction_date = datetime(2025, 1, 1) + timedelta(
        days=random.randint(0, 364)
    )

    transaction_type = random.choice(transaction_types)

    amount = round(random.uniform(100, 50000), 2)

    balance_after_transaction = round(
        random.uniform(1000, 500000), 2
    )

    transactions.append([
        f"T{i:05d}",
        account_id,
        transaction_date.strftime("%Y-%m-%d"),
        transaction_type,
        amount,
        balance_after_transaction
    ])

transactions_df = pd.DataFrame(
    transactions,
    columns=[
        "transaction_id",
        "account_id",
        "transaction_date",
        "transaction_type",
        "amount",
        "balance_after_transaction"
    ]
)

# Save inside Data folder
transactions_df.to_csv("../Data/Transactions.csv", index=False)

print("Transactions.csv created successfully!")
print("Rows:", len(transactions_df))
print(transactions_df.head())
```

```python
import pandas as pd
import random
from datetime import datetime, timedelta

random.seed(42)

# Load existing customers
customers = pd.read_csv("../Data/Customers.csv")

loans = []

loan_types = ["Personal", "Home", "Auto", "Education", "Business"]
loan_statuses = ["Active", "Closed", "Default"]

for i in range(1, 2001):

    customer_id = random.choice(customers["customer_id"].tolist())

    loan_type = random.choice(loan_types)

    loan_amount = round(random.uniform(50000, 2500000), 2)

    interest_rate = round(random.uniform(7, 18), 2)

    loan_date = datetime(2024, 1, 1) + timedelta(
        days=random.randint(0, 730)
    )

    loan_status = random.choices(
        loan_statuses,
        weights=[60, 30, 10]
    )[0]

    outstanding_amount = round(
        loan_amount * random.uniform(0, 0.9), 2
    )

    loans.append([
        f"L{i:05d}",
        customer_id,
        loan_type,
        loan_amount,
        interest_rate,
        loan_date.strftime("%Y-%m-%d"),
        loan_status,
        outstanding_amount
    ])

loans_df = pd.DataFrame(
    loans,
    columns=[
        "loan_id",
        "customer_id",
        "loan_type",
        "loan_amount",
        "interest_rate",
        "loan_date",
        "loan_status",
        "outstanding_amount"
    ]
)

# Save inside Data folder
loans_df.to_csv("../Data/Loans.csv", index=False)

print("Loans.csv created successfully!")
print("Rows:", len(loans_df))
print(loans_df.head())
```
